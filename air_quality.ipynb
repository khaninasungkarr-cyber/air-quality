a. berdasarkan seluruh data polusi yang ada, stasiun manakah yang memilikirata-rata konsentrasi PM2.5 paling tinggi?
b. bagaimana korelasi antara tingkat temperatur udara (temp) dengan konsentrasi polutan PM2.5?

```python
#2. data wrangling
import glob
import pandas as pd

#ambil semua file dalam folder
path = '*.csv'
file = glob.glob(path)
print('jumlah file yang ditemukan', len(file))

#bikin looping buat baca tiap file, trs dimasukin ke list
daftar_df = []
for i in file:
    df_temp = pd.read_csv(i)
    daftar_df.append(df_temp)
    
#gabungin isi list jadi 1 df besar pake concate
df = pd.concat(daftar_df, ignore_index=True)
print('total baris dan kolom keseluruhan:', df.shape)
df.head()
```

```python
print('jumlah data duplicate sebelum dibersihkan:', df.duplicated().sum())
print('jumlah data missing value sebelum dibersihkan: ')
print(df.isnull().sum())
```

```python
#data dibersihkan
df = df.drop_duplicates() #hapus data duplikat
df = df.dropna() #hapus data kosong

print('data bersih:', df.shape)
print('sisa data kosong:', df.isnull().sum())
```

```python
#pertanyaan pertama
analisis_stasiun = (
    df.groupby('station')['PM2.5']
    .mean()
    .reset_index()
    .sort_values(by='PM2.5', ascending=False))
analisis_stasiun['PM2.5'] = analisis_stasiun['PM2.5'].round(2)
print('rata-rata konsentrasi PM2.5 pertahun')
print(analisis_stasiun)
```

jawaban dari pertanyaan pertama yaitu rata-rata PM2.5 tertinggi diduduki oleh stasiun Dongsi dengan rata-rata 84.93, kemudian untuk rata-rata terendah dengan angka 66.51 diduduki oleh stasiun Dingling.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,6))
sns.barplot(data=analisis_stasiun, x='station', y= 'PM2.5')
plt.title('Diagram rata-rata PM2.5 di tiap stasiun', fontsize = 12, fontweight='bold')
plt.xlabel('nama stasiun')
plt.ylabel('PM2.5')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# pertanyaan kedua
korelasi = df[['TEMP','PM2.5']].corr().loc['TEMP','PM2.5'] # loc untuk mengambil angka tunggal pada baris korelasi

plt.figure(figsize=(12,6))
sns.regplot(data=df, x='TEMP', y='PM2.5')
plt.title('korelasi antara temperature dan konsentrasi PM2.5', fontsize=12)
plt.xlabel('temperatur (C)')
plt.ylabel('PM2.5')
plt.tight_layout()
plt.show()
```

jawaban dari pertanyaan kedua, didapatkan bahwa pada grafik diatas terlihan bahwa temperatur tidak memiliki hubungan terhadap PM2.5. bisa dilihat ketika suhu -20C, dia memiliki konsentrasi polutas hampir mendekati 0. tidak bisa disimpulkan bahwa suhu tinggi menyebabkan banyak polutan maupun sebaliknya.
